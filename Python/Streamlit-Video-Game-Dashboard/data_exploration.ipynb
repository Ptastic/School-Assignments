# Cleaning, searching and shaping data

```python
# Load the datasets
Games_score = pd.read_csv("Resources/games-score.csv")
Consoles_sales = pd.read_csv("Resources/best-selling game consoles.csv")
Game_sales = pd.read_csv("Resources/video games sales.csv")

# Quick look at the data
print("Games_score shape:", Games_score.shape)
print("Consoles_sales shape:", Consoles_sales.shape)
print("Game_sales shape:", Game_sales.shape)
```

Testing, analyzing

```python
# Clean Games_score
Games_score["user_score"] = pd.to_numeric(Games_score["user_score"], errors="coerce") * 10  
Games_score["release_date"] = pd.to_datetime(Games_score["release_date"])

# Clean Game_sales
Game_sales["Year"] = Game_sales["Year"].fillna(Game_sales["Year"].median())
Game_sales["Publisher"] = Game_sales["Publisher"].fillna("Unknown")

# Check for missing values
print("Missing values in Games_score:")
print(Games_score[["user_score", "metascore"]].isna().sum())
print("\nMissing values in Game_sales:")
print(Game_sales[["Name", "Platform", "Global_Sales"]].isna().sum())
```

```python
# Normalize platform names for merging
platform_normalization = {
    "PS": "PlayStation", "PS2": "PlayStation 2", "PS3": "PlayStation 3", "PS4": "PlayStation 4",
    "XB": "Xbox", "X360": "Xbox 360", "XOne": "Xbox One",
    "NES": "NES", "SNES": "SNES", "Wii": "Wii", "Switch": "Nintendo Switch",
    "DC": "Dreamcast", "Wii U": "Wii U", "WiiU": "Wii U",
    "GC": "GameCube", "3DS": "Nintendo 3DS",
    "GB": "Game Boy", "GBA": "Game Boy Advance", "DS": "Nintendo DS", "N64": "Nintendo 64",
    "PSP": "PlayStation Portable", "GBC": "Game Boy Color", "DSI": "Nintendo DSi",
    "GEN": "Sega Genesis", "SAT": "Sega Saturn", "S32X": "Sega 32X", "SCD": "Sega CD",
    "PC": "PC"
}
Game_sales["Platform_normalized"] = Game_sales["Platform"].map(platform_normalization)
Games_score["platform_normalized"] = Games_score["platform"].map(lambda x: platform_normalization.get(x, x))

# Merge the datasets
merged_games = pd.merge(Game_sales, Games_score, left_on=["Name", "Platform_normalized"], 
                        right_on=["title", "platform_normalized"], how="left")

# Check the merge
print("Merged games shape:", merged_games.shape)
print("Missing metascore after merge:", merged_games["metascore"].isna().sum())
print("Missing user_score after merge:", merged_games["user_score"].isna().sum())

# Check Wii U games specifically
wiiu_games = merged_games[merged_games["Platform"].str.contains("WiiU", na=False)]
print("Number of Wii U games:", len(wiiu_games))
print("Wii U games with metascore:", wiiu_games["metascore"].notna().sum())
print("Wii U games with user_score:", wiiu_games["user_score"].notna().sum())
print("Sample of Wii U games:")
print(wiiu_games[["title", "platform", "metascore", "user_score"]].head())
```

```python
# Map platforms to manufacturers
manufacturer_map = {
    "PS": "Sony", "PS2": "Sony", "PS3": "Sony", "PS4": "Sony", "PSP": "Sony",
    "XB": "Microsoft", "X360": "Microsoft", "XOne": "Microsoft",
    "NES": "Nintendo", "SNES": "Nintendo", "Wii": "Nintendo", "Switch": "Nintendo",
    "DC": "Sega", "Wii U": "Nintendo", "GC": "Nintendo", "3DS": "Nintendo",
    "GB": "Nintendo", "GBA": "Nintendo", "DS": "Nintendo", "N64": "Nintendo",
    "GBC": "Nintendo", "DSI": "Nintendo",
    "GEN": "Sega", "SAT": "Sega", "S32X": "Sega", "SCD": "Sega",
    "PC": "Others"
}
merged_games["manufacturer"] = merged_games["Platform"].map(manufacturer_map).fillna("Others")
merged_games = merged_games.drop_duplicates(subset=["Name", "Platform"])
merged_games["avg_score"] = merged_games[["metascore", "user_score"]].mean(axis=1)

# Check the distribution of manufacturers
print("Number of games per manufacturer:")
print(merged_games["manufacturer"].value_counts())
```

```python
# Define color schemes for experimentation
manufacturer_colors = {
    "Nintendo": "#1f77b4",  # Blue
    "Sony": "#2ca02c",      # Green
    "Microsoft": "#d62728", # Red
    "Others": "#7f7f7f",    # Gray
    "Sega": "#ff7f0e"       # Orange
}

# Choropleth map color scale (shades of blue)
choropleth_color_scale = "PiYg"

# Filter for Others and try a scatterplot
filtered_games = merged_games[merged_games["manufacturer"] == "Others"]
if filtered_games["metascore"].notna().sum() == 0:
    print("No score data available for Others.")
else:
    fig_scores = px.scatter(filtered_games, x="metascore", y="user_score", hover_data=["Name"], 
                            title="Metascore vs User Score (0-100)",
                            color="manufacturer", color_discrete_map=manufacturer_colors)
    fig_scores.show()

# Top 15 games by avg_score
top_15_scores = filtered_games[filtered_games["avg_score"].notna()].nlargest(15, "avg_score")
fig_top_scores = px.bar(top_15_scores, x="Name", y="avg_score", title="Top 15 Games (Average Score 0-100)",
                        color="manufacturer", color_discrete_map=manufacturer_colors)
fig_top_scores.show()

# Bottom 15 games by avg_score
bottom_15_scores = filtered_games[filtered_games["avg_score"].notna()].nsmallest(15, "avg_score")
fig_bottom_scores = px.bar(bottom_15_scores, x="Name", y="avg_score", title="Bottom 15 Games (Average Score 0-100)",
                           color="manufacturer", color_discrete_map=manufacturer_colors)
fig_bottom_scores.show()

# Shared page visualizations (all consoles)
# Top games (treemap)
top_games_all = merged_games.nlargest(10, "Global_Sales")
fig_top_games_all = px.treemap(top_games_all, path=["manufacturer", "Name"], values="Global_Sales",
                               title="Top 10 Games by Sales (All Consoles)",
                               color="manufacturer", color_discrete_map=manufacturer_colors)
fig_top_games_all.show()

# Console sales (pie chart)
console_sales_by_company = Consoles_sales.groupby("Company")["Units sold (million)"].sum().reset_index()
fig_console_sales_all = px.pie(console_sales_by_company, names="Company", values="Units sold (million)",
                               title="Console Sales by Company (All Consoles)",
                               color="Company", color_discrete_map=manufacturer_colors)
fig_console_sales_all.show()

# Scores (box plot)
fig_scores_all = px.box(merged_games[merged_games["avg_score"].notna()], x="manufacturer", y="avg_score",
                        title="Score Distribution by Manufacturer (All Consoles)",
                        color="manufacturer", color_discrete_map=manufacturer_colors)
fig_scores_all.show()

# Choropleth map for regional sales
# Aggregate regional sales
regional_sales = {
    "Region": ["NA", "EU", "JP"],
    "Sales": [
        merged_games["NA_Sales"].sum(),
        merged_games["EU_Sales"].sum(),
        merged_games["JP_Sales"].sum()
    ]
}
regional_sales_df = pd.DataFrame(regional_sales)

# Map regions to countries (simplified)
region_to_countries = {
    "NA": ["USA", "CAN", "MEX"],
    "EU": ["DEU", "FRA", "ITA", "ESP", "SWE", "DNK", "FIN", "POL", "BEL", "GRC", "ROU", "IRL", "EST", "LTU", 
            "NLD", "CHE", "AUT", "HUN", "SVK", "SVN", "CZE", "HRV", "PRT", "LUX", "MLT", "CYP", "LVA", "BGR",
            "BLR","UKR","ISL","GBR","NOR","ALB","MKD","SRB","BIH","MNE","MLD","MDA"],  
    "JP": ["JPN"]
}

# Create a DataFrame for the choropleth map
choropleth_data = []
for _, row in regional_sales_df.iterrows():
    region = row["Region"]
    sales = row["Sales"]
    for country in region_to_countries[region]:
        choropleth_data.append({"Country": country, "Sales": sales, "Region": region})
choropleth_df = pd.DataFrame(choropleth_data)

# Create the choropleth map
fig_choropleth = px.choropleth(choropleth_df, locations="Country", locationmode="ISO-3",
                               color="Sales", hover_name="Country", hover_data=["Region"],
                               title="Regional Sales (Millions)",
                               color_continuous_scale=choropleth_color_scale)
fig_choropleth.show()

# Mock game search
search_term = "Mario"  # You can change this to test different games
search_results = merged_games[merged_games["Name"].str.contains(search_term, case=False, na=False)]
print(f"Search results for '{search_term}':")
#print(search_results[["Name", "Platform", "Manufacturer", "Global_Sales", "Metascore", "User_score", "Avg score"]].head(10))
```
